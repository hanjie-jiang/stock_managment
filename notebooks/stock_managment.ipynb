# Family Stock Research Toolkit

Decision-support workspace covering: research, buy/sell timing signals, cross-stock comparison, risk scanning, long-term value screening, fundamentals, and quarterly report reading -- for both US and China-listed stocks (A-shares, HK, ADRs).

Powered by `stock_toolkit.py` (uses `yfinance`, no API key needed). Everything below is **data + heuristics, not investment advice** -- verify against actual filings before acting.

Edit `WATCHLIST` in the next cell to track your own tickers. Ticker format:
- US: `AAPL`, `MSFT`
- China A-share: `600519.SS` (Shanghai), `000858.SZ` (Shenzhen)
- Hong Kong: `0700.HK`
- China ADR (trades in US): `BABA`, `JD`, `PDD`

```python
import sys
from pathlib import Path

# Allow importing stock_toolkit.py from the project root when this notebook
# runs from inside notebooks/
sys.path.insert(0, str(Path.cwd().parent))

import stock_toolkit as st

# Your watchlist -- mix of US and China tickers as an example
WATCHLIST = ["AAPL", "MSFT", "600519.SS", "0700.HK", "BABA"]
```

## 1. Research

Company profile, sector/industry, and recent headlines for a single ticker.

```python
st.print_research("AAPL")
```

## 2 & 3. Is it worth buying in? Is it a good time to sell?

Combines valuation (vs. analyst targets, forward vs trailing P/E), momentum (price vs 50/200-day moving averages, RSI), and recent growth trends into a bull/bear signal list with a lean (not a verdict).

```python
for symbol in WATCHLIST:
    st.print_buy_sell_signal(symbol)
    print()
```

## 4. Compare stocks side by side

Valuation, profitability, growth, leverage, and analyst view in one table.

```python
st.compare_stocks(WATCHLIST)
```

## 5. Risk scan

Beta, annualized volatility, max drawdown, leverage, liquidity, and short interest -- flagged when they cross common risk thresholds.

```python
for symbol in WATCHLIST:
    st.print_risk_scan(symbol)
    print()
```

## 6. Long-term / value-investing checklist

A simple Buffett-style checklist: ROE, margins, free cash flow, leverage, revenue/earnings growth consistency, and liquidity cushion.

```python
for symbol in WATCHLIST:
    st.print_long_term_value_score(symbol)
    print()
```

## 7. Fundamentals -- full financial statements

Full income statement, balance sheet, and cash flow (annual or quarterly) for deep-dive fundamental analysis. Pick one ticker to inspect.

```python
FOCUS_SYMBOL = "AAPL"

data = st.fundamentals(FOCUS_SYMBOL, freq="annual")  # or freq="quarterly"
data["income_statement"]
```

```python
data["balance_sheet"]
```

```python
data["cashflow"]
```

## 8. Quarterly report reader (US 10-Q & China quarterly filings)

Latest quarter's key line items with quarter-over-quarter and year-over-year % change, so you can quickly see whether a new earnings report is actually good or bad relative to what came before.

```python
for symbol in WATCHLIST:
    st.print_quarterly_report_summary(symbol)
    print()
```

### Notes & limitations
- Data comes from Yahoo Finance via `yfinance`. It's generally reliable but occasionally has gaps for China A-shares (e.g. missing EPS in some quarters) -- cross-check anything decision-critical against the company's actual filing (SEC EDGAR for US, cninfo.com.cn / HKEXnews for China).
- The buy/sell "lean" and the value-investing checklist are simple rule-based heuristics, not predictions or recommendations -- they're meant to surface signals worth digging into, not to replace judgment.
